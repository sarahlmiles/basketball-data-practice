```python
games = {
    "Player": ["Brunson", "Brunson", "SGA", "SGA", "Tatum", "Tatum", "Jokic",
               "Jokic", "Curry", "Curry"],
    "Game": [1, 2, 1, 2, 1, 2, 1, 2, 1, 2],
    "PTS": [26, 31, 33, 28, 27, 24, 30, 35, 25, 29],
    "AST": [7, 9, 6, 8, 6, 5, 10, 12, 6, 8],
    "REB": [3, 4, 5, 6, 9, 7, 13, 11, 4, 5]
}
```
